```python
# fetch dataset 
secondary_mushroom = fetch_ucirepo(id=848)
```

```python
# data (as pandas dataframes) 
X = secondary_mushroom.data.features 
y = secondary_mushroom.data.targets
```

### VISUALIZATION OF THE DATA

```python
sns.set(style="white")

# Generate a random univariate dataset
rs = np.random.RandomState(10)
d = rs.normal(size=100)

# Plot a simple histogram and kde
sns.histplot(d, kde=True, color="m")
```

```python
mush_feat = ['cap-diameter','stem-height','stem-width']

var_color = ["#FF5733", "#33FF57", "#3357FF"]

X_melted = X[mush_feat].melt(var_name="Variable", value_name="Value")
fig, ax = plt.subplots(1,4, figsize=(15,8)) 

#axis-plot
sns.boxplot(data=X_melted, x="Variable",hue="Variable",y="Value" , ax=ax[0], palette='tab10')
ax[0].set_title("AxisPlot of Mush Features")

#scartter-plot
sns.scatterplot(data=X_melted, x="Variable",hue="Variable",y="Value" , ax=ax[1], palette='tab10')
ax[1].set_title("Scatter plot")

#surface-plot
sns.histplot(data=X_melted, x="Variable",hue="Variable",y="Value" , ax=ax[2], palette='deep')
ax[2].set_title('Mushroom subplots') 

#for displot finding the continous plot of the Variable
for i, feat in enumerate(mush_feat):
    sns.kdeplot(data=X[feat], label=feat, ax=ax[3], fill=True, color=var_color[i])
ax[3].set_title("DisPlot of Mush Features")
ax[3].set_xlabel("Value") 
ax[3].legend(title="Variable")

#fig.canvas.draw() 
plt.tight_layout()
plt.show() 
```

```python
X = pd.get_dummies(X)
```

The connection of the import libraries pd to frames data for the feautures and the target data for the secondary mushroom.This will then get or create a dummies from X.

The models to used.This model will test the case of each models used by the following method and process(ed).

```python
model = Perceptron()
# model = svm.SVC()
# model = KNeighborsClassifier(n_neighbors=1)
#model = GaussianNB()
```

the trained data will be then split based on their test_size to get from the whole set of data with a random state to get the given describe data in shuffle or random way.

```python
X_training, X_testing, y_training, y_testing = train_test_split(
    X, y, test_size=0.10, random_state=42
)
y_training = y_training.values.ravel()
y_testing = y_testing.values.ravel()

scaler = StandardScaler()
X_training = scaler.fit_transform(X_training)
X_testing = scaler.fit_transform(X_testing)
```

the model libraries of scaler from sk-learn is then used to normalize or standardize the included training and testing data in X scale or X group.

the model libraries of ravel is then apply on the included trained and tested data in y group or y scale.

```python
# Fit model
model.fit(X_training, y_training)
```

The dataset is then takes into training which in this case has X and y or we should say two arrays or two groups of trained data fit.

```python
# Make predictions on the testing set
predictions = model.predict(X_testing)
```

This will perform a prediction of the test and usually the ,predict only accepts single input in which  in this case it only has (X_testing).

```python
# Compute how well we performed
correct = (y_testing == predictions).sum()
incorrect = (y_testing != predictions).sum()
total = len(predictions)
```

```python
# Print results
print(f"Results for model {type(model).__name__}")
print(f"Correct: {correct}")
print(f"Incorrect: {incorrect}")
print(f"Accuracy: {100 * correct / total:.2f}%")
```
